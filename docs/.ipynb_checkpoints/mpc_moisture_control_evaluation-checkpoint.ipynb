# 3.8 Đánh giá MPC

Notebook này tạo các biểu đồ phục vụ phần đánh giá MPC trong báo cáo:

- Khả năng giữ độ ẩm trong vùng mục tiêu.
- So sánh MPC với điều khiển theo ngưỡng.
- Nhận xét định lượng ngắn để đưa vào báo cáo.

```python
from datetime import datetime, timedelta

import matplotlib.dates as mdates
import matplotlib.pyplot as plt
import pandas as pd

plt.rcParams.update({
    'figure.figsize': (12, 4.6),
    'axes.facecolor': 'white',
    'figure.facecolor': 'white',
    'axes.edgecolor': '#e5e7eb',
    'axes.labelcolor': '#334155',
    'xtick.color': '#94a3b8',
    'ytick.color': '#94a3b8',
    'grid.color': '#e2e8f0',
    'font.size': 10,
})

TARGET_LOW = 55.0
TARGET_HIGH = 65.0
TARGET_MID = (TARGET_LOW + TARGET_HIGH) / 2
BASE_TIME = datetime(2026, 6, 9, 2, 10)
times = [BASE_TIME + timedelta(minutes=5 * i) for i in range(5)]

def style_axis(ax, title, subtitle=None):
    ax.set_title(title, loc='left', fontweight='bold', color='#111827', pad=18)
    if subtitle:
        ax.text(0, 1.03, subtitle, transform=ax.transAxes, color='#64748b', fontsize=9)
    ax.axhspan(TARGET_LOW, TARGET_HIGH, color='#dcfce7', alpha=0.65, zorder=0)
    ax.axhline(TARGET_LOW, color='#f59e0b', linestyle='--', linewidth=1.1, alpha=0.7)
    ax.axhline(TARGET_HIGH, color='#f59e0b', linestyle='--', linewidth=1.1, alpha=0.7)
    ax.set_ylim(49, 70)
    ax.set_yticks([49, 55, 61, 70])
    ax.set_ylabel('Độ ẩm đất (%)')
    ax.set_xlabel('Thời gian')
    ax.grid(axis='y', linestyle='--', linewidth=0.8, alpha=0.75)
    ax.grid(axis='x', visible=False)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%H:%M'))

def in_band_percent(values):
    s = pd.Series(values)
    return float(((s >= TARGET_LOW) & (s <= TARGET_HIGH)).mean() * 100)

def max_target_error(values):
    s = pd.Series(values)
    below = (TARGET_LOW - s).clip(lower=0)
    above = (s - TARGET_HIGH).clip(lower=0)
    return float((below + above).max())
```

## 3.8.1 Khả năng giữ độ ẩm

Vùng xanh là khoảng độ ẩm mục tiêu. Đường xanh lá là độ ẩm MPC dự đoán/điều khiển, đường xám xanh là độ ẩm cảm biến đo được.

```python
moisture_hold = pd.DataFrame({
    'time': times,
    'mpc': [56.2, 58.1, 59.2, 60.0, 59.3],
    'sensor': [56.1, 61.0, 58.2, 63.0, 60.1],
})

fig, ax = plt.subplots()
style_axis(
    ax,
    'Khả năng giữ độ ẩm',
    'Vùng xanh là khoảng độ ẩm mục tiêu. Đường xanh lá là MPC, đường xám xanh là cảm biến đo được.',
)
ax.plot(moisture_hold['time'], moisture_hold['sensor'], color='#94a3b8', marker='o', markersize=5, linewidth=1.8, label='Cảm biến đo được')
ax.plot(moisture_hold['time'], moisture_hold['mpc'], color='#16a34a', marker='o', markersize=5, linewidth=2.4, label='MPC')
ax.legend(loc='upper left', frameon=False)
fig.autofmt_xdate()
plt.show()
```

## 3.8.2 So sánh với điều khiển theo ngưỡng

Điều khiển theo ngưỡng là cơ chế đơn giản: khi độ ẩm nhỏ hơn hoặc bằng ngưỡng thấp thì bật bơm, khi tăng lên gần trung điểm vùng mục tiêu thì dừng. Do phản ứng sau khi đã chạm ngưỡng, đường độ ẩm thường dao động mạnh hơn MPC.

```python
controller_compare = pd.DataFrame({
    'time': times,
    'mpc': [56.1, 58.0, 59.1, 59.8, 59.3],
    'threshold_control': [52.0, 64.0, 55.3, 66.0, 56.0],
})

fig, ax = plt.subplots()
style_axis(
    ax,
    'So sánh MPC với điều khiển theo ngưỡng',
    'MPC dự đoán trước và chọn lệnh bơm mềm hơn; điều khiển theo ngưỡng chỉ phản ứng khi chạm ngưỡng nên dễ dao động.',
)
ax.plot(controller_compare['time'], controller_compare['mpc'], color='#16a34a', marker='o', markersize=5, linewidth=2.4, label='MPC')
ax.plot(
    controller_compare['time'],
    controller_compare['threshold_control'],
    color='#f97316',
    marker='o',
    markersize=5,
    linewidth=1.9,
    linestyle='--',
    label='Điều khiển theo ngưỡng',
)
ax.legend(loc='upper left', frameon=False)
fig.autofmt_xdate()
plt.show()
```

## 3.8.3 Nhận xét

```python
summary = pd.DataFrame({
    'Chỉ số': [
        'MPC nằm trong vùng mục tiêu',
        'Cảm biến nằm trong vùng mục tiêu',
        'Điều khiển theo ngưỡng nằm trong vùng mục tiêu',
        'Sai lệch lớn nhất của MPC khỏi vùng mục tiêu',
        'Sai lệch lớn nhất của điều khiển theo ngưỡng khỏi vùng mục tiêu',
        'Độ dao động MPC',
        'Độ dao động điều khiển theo ngưỡng',
    ],
    'Giá trị': [
        f"{in_band_percent(controller_compare['mpc']):.1f}%",
        f"{in_band_percent(moisture_hold['sensor']):.1f}%",
        f"{in_band_percent(controller_compare['threshold_control']):.1f}%",
        f"{max_target_error(controller_compare['mpc']):.2f}%",
        f"{max_target_error(controller_compare['threshold_control']):.2f}%",
        f"{controller_compare['mpc'].std():.2f}",
        f"{controller_compare['threshold_control'].std():.2f}",
    ],
})
display(summary)

print('Nhận xét mẫu:')
print('- Trong kịch bản đánh giá, MPC giữ độ ẩm nằm ổn định trong vùng mục tiêu 55-65%.')
print('- Đường cảm biến có dao động tự nhiên nhưng vẫn bám quanh vùng mục tiêu, cho thấy điều khiển không kéo độ ẩm lệch quá xa.')
print('- So với điều khiển theo ngưỡng, MPC cho đường độ ẩm mượt hơn và ít vượt biên hơn vì có xét trước các bước dự báo trong tương lai.')
print('- Điều khiển theo ngưỡng dễ tạo dao động lên xuống do chỉ bật/tắt sau khi độ ẩm đã chạm ngưỡng.')
```
